# Sales analysis

Quarterly sales exploration (now with Q5).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
def summarize(frame):
    """Describe the numeric columns."""
    return frame.describe()
```

```python
rows = [("Q1", 3), ("Q2", 5), ("Q3", 4), ("Q4", 6), ("Q5", 5)]
df = pd.DataFrame(rows, columns=["quarter", "sales"])
df.head()
```

```python
total = df["sales"].sum()
print("total sales:", total)
```
